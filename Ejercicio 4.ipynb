# Calculadora geométrica básica

## Area de un rectángulo

```python
base = float (input ('Captura la base: '))
altura = float (input ('Captura la altura: '))
area = base * altura
perimetro = (base+altura)*2
print ('El área del rectángulo es : ', area)
print ('El perímetro del rectángulo es : ', perimetro)
```

## Area de un triángulo equilatero

```python
base = float (input ('Captura la base: '))
area = base * base / 2
perimetro = base * 3
print ('El área del triangulo es : ', area)
print ('El perímetro del triángulo es : ', perimetro)
```

## Area y perimetro de un círculo

```python
radio = float (input ('Captura el valor del radio: '))
area = 3.1416 * (radio**2)
perimetro = 3.1416 * (radio*2)
print ('El área del círculo es : ', area)
print ('El perímetro del círculo es : ', perimetro)
```
