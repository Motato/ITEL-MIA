# **Manejo de listas con valores numéricos y valores textuales**

```python
letras = ["a","b","c","d","e","f","g"]
print(letras)
```

```python
numeros =[1,2,8,5,6]
print(numeros)
```

#  **Reemplazando algunos valores en las listas**

```python
letras[2:5]=[ ] #se eliminan las letras de la posición 2 a la 4 la c,d y e
print(letras)
```

# **Sustituyendo letras**

```python
letras[ 1:2 ] = ['B','F']
print(letras)
```
