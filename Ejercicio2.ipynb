# Operaciones con un número

```python
numero = int (input ('Captura un número:'))
doble = numero * 2
triple = numero * 3
raiz = numero ** 0.5
print('El doble del numero que ingresaste es : ', doble)
print('El triple del numero que ingresaste es : ', triple)
print('La raíz cuadrada del número que ingresaste es : ', raiz)
```
