# Conversión de grados Celsius a Farenheit y Kelvin

```python
celsius = float (input ('Captura un valor en grados Celsius: '))
farenheit = (celsius * 9/5) +32
kelvin = celsius + 273.5
print('El equivalente de los grados Celsius que ingresaste convertidos a Farenheit es : ', farenheit)
print('El equivalente de los grados Celsius que ingresaste convertidos a kelvin es : ', kelvin)
```
